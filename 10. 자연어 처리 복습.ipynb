```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding
import numpy as np

text = '커피 한잔 어때'

# 1. Tokenizer 객체 생성
tokenizer = Tokenizer()

# 2. 텍스트 학습
tokenizer.fit_on_texts([text])  # 여러개 넣을 수 있으므로 리스트로 전달

# 3. tokenizer 객체 출력
print("tokenizer: ", tokenizer)

# 4. 텍스트를 시퀀스로 변환
sequences = tokenizer.texts_to_sequences([text])
print("sequences: ", sequences)

# 5. 단어 인덱스 출력
word_index = tokenizer.word_index
print("word_index: ", word_index)

# 6. 인덱스 -> 단어 맵핑 출력
index_word = tokenizer.index_word
print("index_word: ", index_word)
```

**단어 빈도 수 세기**

```python
docs = ['먼저 텍스트의 각 단어를 나누어 토큰화합니다.',
       '텍스트의 단어로 토큰화해야 딥러닝에서 인식됩니다.',
       '토큰화한 결과는 딥러닝에서 사용할 수 있습니다.',
       ]

tokenizer = Tokenizer()
tokenizer.fit_on_texts(docs)

# 출력되는 순서는 랜덤
print("단어 카운트: ", tokenizer.word_counts)
print("문장 카운트: ", tokenizer.document_count)
print("각 단어가 몇 개의 문장에 포함되어 있는가: ", tokenizer.word_docs)
print("각 단어에 매겨진 인덱스 값: ",  tokenizer.word_index)

# word_counts: 모든 문장에서 각 단어가 등장한 총 횟수
# document_count: 전체 문장의 수
# word_docs: 각 단어가 등장한 문장 수
# word_index: 빈도 높은 단어 순서대로 매겨진 인덱스 딕셔너리
```

```python
texts = ['커피 한잔 어때', '오늘 날씨 참 좋네', '옷이 어울려요']

# 1. 토크나이저 생성 및 학습
tokenizer = Tokenizer()
tokenizer.fit_on_texts(texts)

# 2. 시퀀스로 변환
sequences = tokenizer.texts_to_sequences(texts)
print(sequences)

# 3. 단어 인덱스 출력
word_index = tokenizer.word_index
print("\n단어 인덱스: ", word_index)

# 4. 패딩 처리 (길이를 4로 통일)
print("\n시퀀스: ", sequences)
padded_sequences = pad_sequences(sequences, 4)
print("\n패딩된 시퀀스:\n", padded_sequences)
```

```python
model = Sequential()
model.add(Embedding(input_dim = len(word_index)+1, output_dim = 5))
# Embedding: 숫자를 벡터로 바꿈
# input_dim: 단어 집합의 크기 → len(word_index)+1  (+1은 패딩 토큰 0을 위한 자리)
# output_dim: 각 단어를 임베딩할 벡터 차원 수 (여기서는 5차원)


embedding_output = model.predict(padded_sequences) # padded_sequences를 모델에 통과시켜 임베딩된 결과 추출
embedding_output
```
